```python
import asyncio

import crewai
import mlflow

from router_flow import AgricultureRouterFlow

print("CrewAI version:", crewai.__version__)
```

```python
# Optional: MLflow tracking, in its own experiment so runs don't mix with the old crew's
mlflow.set_tracking_uri("sqlite:///mlflow.db")
mlflow.set_experiment("Agriculture Router Flow")
mlflow.crewai.autolog()
```

```python

flow = AgricultureRouterFlow()

def run_flow(user_question):
    return flow.kickoff(inputs={"user_question": user_question})
```

```python
user_question = "How has cotton yield in Nanded changed over the last 10 years, and what pest management practices are recommended for cotton?"

result = await asyncio.to_thread(run_flow, user_question)
print(result)
```
