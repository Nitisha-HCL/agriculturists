```python
from pathlib import Path
from crewai.project import load_crew
import crewai
import mlflow
import asyncio
print("CrewAI version:", crewai.__version__)
```

```python
# Optional: Set a tracking URI and an experiment
mlflow.set_tracking_uri("sqlite:///mlflow.db")
mlflow.set_experiment("Agriculture Sequential Basic")
```

```python
mlflow.crewai.autolog()
```

```python

crew, default_inputs = load_crew(
    Path("crew.jsonc")
)

```

```python
def run_crew():
    return crew.kickoff(
    inputs={
        "user_question": "How has cotton yield in distict Nanded changed over the last 10 years?",

    }
)
```

```python
result = await asyncio.to_thread(run_crew)
print(result)
```
